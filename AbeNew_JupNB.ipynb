4. Display the plot

```python
# Step 1: Import matplotlib
import matplotlib.pyplot as plt
```

## Step 2: Define Data

We'll define the data points that we want to plot. The x-values represent time (in arbitrary units), and the y-values represent some measured values.

```python
import pandas as pd
from io import StringIO

# (1) Paste your full data here, replacing the "..." block with all lines
raw_data = """Date,PCP
1/1/1991,0
1/2/1991,0
1/3/1991,0
1/4/1991,0
1/5/1991,0
1/6/1991,0
1/7/1991,13.1227
...
12/31/1991,0
"""

# (2) Read it into pandas
df = pd.read_csv(
    StringIO(raw_data),
    parse_dates=['Date'],       # automatically convert "Date" column to datetime
    dayfirst=False              # our format is M/D/YYYY
)

# (3) Make Date the index
df.set_index('Date', inplace=True)

# Quick check
df.head()
```

```python
# e.g. a simple plot
df.PCP.plot(title="Daily Precipitation, 1991", figsize=(10,4))
df = pd.read_csv("your_precip_1991.csv", parse_dates=['Date'], index_col='Date')

```
